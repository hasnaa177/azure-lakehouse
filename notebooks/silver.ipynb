```python
df = (spark.read
      .option("header", True)
      .option("multiLine", True)
      .option("escape", '"')
      .csv(bronze_path_in))

df.printSchema()
print("Lignes lues :", df.count())
display(df.limit(10))
```

noms de colonnes propres

```python
import re
from pyspark.sql import functions as F

df = df.toDF(*[re.sub(r"[^0-9a-zA-Z]+", "_", c.strip()).strip("_").lower() for c in df.columns])
```

typage

```python
def to_double(col):
    return F.expr(f"try_cast(regexp_replace({col}, '[^0-9.]', '') as double)")

def to_int(col):
    return F.expr(f"try_cast(regexp_replace({col}, '[^0-9]', '') as int)")

df_clean = (df
    .withColumn("discounted_price",    to_double("discounted_price"))
    .withColumn("actual_price",        to_double("actual_price"))
    .withColumn("discount_percentage", to_double("discount_percentage"))
    .withColumn("rating",              to_double("rating"))
    .withColumn("rating_count",        to_int("rating_count"))
    .withColumn("main_category",       F.split("category", r"\|").getItem(0))
)

df_clean.printSchema()
```

doublons

```python
before = df_clean.count()
df_clean = df_clean.dropDuplicates()
print("Doublons exacts supprimés :", before - df_clean.count())

print("Lignes :", df_clean.count())
print("Produits distincts :", df_clean.select("product_id").distinct().count())
# df_clean = df_clean.dropDuplicates(["product_id"])   # décommente pour 1 ligne par produit
```

valeurs manquantes

```python
display(df_clean.select([F.sum(F.col(c).isNull().cast("int")).alias(c) for c in df_clean.columns]))

df_clean = df_clean.dropna(subset=["product_id", "discounted_price", "actual_price"])

avg_rating = df_clean.select(F.avg("rating")).first()[0]
df_clean = df_clean.fillna({
    "rating": round(avg_rating, 1),
    "rating_count": 0,
    "about_product": "Inconnu"
})

print("Lignes après nettoyage :", df_clean.count())
display(df_clean.limit(10))
```

écriture en Delta dans silver

```python
(df_clean.write
    .format("delta")
    .mode("overwrite")
    .save(silver_path_out))
```

vérification

```python
df_check = spark.read.format("delta").load(silver_path_out)
print("Lignes Delta :", df_check.count())
df_check.printSchema()
display(df_check.limit(5))
```
